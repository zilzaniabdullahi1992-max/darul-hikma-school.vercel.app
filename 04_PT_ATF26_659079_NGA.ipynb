### 1.1 Zero-Shot Prompting

```python
# Example prompts for digital twin
zero_shot_examples = [
    {
        'task': 'activity_prediction',
        'prompt': 'Predict the next activity: I just finished work, had dinner, and now...',
        'expected': 'leisure/relaxation activity'
    },
    {
        'task': 'preference_extraction',
        'prompt': 'Extract preferences from: I love outdoor activities but hate crowded places',
        'expected': {'likes': ['outdoor', 'quiet'], 'dislikes': ['crowds']}
    }
]

print('Zero-Shot Examples:')
for ex in zero_shot_examples:
    print(f"\nTask: {ex['task']}")
    print(f"Prompt: {ex['prompt']}")
    print(f"Expected: {ex['expected']}")
```

### 🎯 Exercise 1.1 (Easy): Create Your Prompts

**Task**: Design prompts for your digital twin use case.

```python
# YOUR CODE HERE
TWIN_PROFILE = """
Name: Abdullahi Zilzani
Role: Education Officer and Examination Officer, Lugu Marina Primary School, Tsafe, Zamfara State, Nigeria
Education: B.Sc. Computer Science (ABU Zaria); currently pursuing M.Sc. Artificial Intelligence (ABU Zaria)
Work: school management systems, mobile and web apps (Flutter, Node/Express, PostgreSQL), primary English textbook
Business: Fadama Guys Livestock (goats, rams, cows) and its social media
Languages: English and Hausa
Style: direct, concise, actionable
"""

SYSTEM_PROMPT = f"""You are the digital twin of Abdullahi Zilzani.
Speak in first person, in his voice: direct, concise, practical.

PROFILE:
{TWIN_PROFILE}

RULES:
1. Only state facts that are in the profile or the conversation. If unsure, say "I'd need to confirm that."
2. Reply in the language the sender uses (English or Hausa).
3. Never agree to payments, prices, deadlines, or livestock sales. Say "Let me confirm and get back to you."
4. For anything sensitive or high-stakes, draft a reply and flag it for my review.
"""

PROMPTS = {
    # 1. Q&A about me
    "about_me": "Someone asks: '{question}'. Answer as me in 2-3 sentences using only my profile.",

    # 2. Message drafting (bilingual)
    "draft_message": """Draft a {channel} message to {recipient} about: {topic}.
Tone: {tone}. Language: {language}. Max {max_words} words. End with one clear next step.""",

    # 3. Livestock business
    "livestock_post": """Write a social media post for Fadama Guys Livestock about: {topic}.
Audience: buyers in northern Nigeria. Include a call to action. Give 2 versions: English and Hausa.""",

    # 4. School work
    "school_task": """Task for my role as Examination Officer: {task}.
Give a short checklist (max 6 items) I can act on today.""",

    # 5. Software / project help
    "project_help": """Project: {project}. Stack: {stack}. Problem: {problem}.
Give the likely cause, the fix, and one way to test it. No long explanations.""",

    # 6. M.Sc. study helper
    "study_helper": """Explain '{concept}' for an M.Sc. AI student in under 150 words,
then give one small Python example and one question to test my understanding.""",
}

def build_prompt(name, **kwargs):
    return SYSTEM_PROMPT + "\n\nUSER REQUEST:\n" + PROMPTS[name].format(**kwargs)

# Test
print(build_prompt("draft_message", channel="WhatsApp", recipient="a parent",
                   topic="exam timetable change", tone="polite, formal",
                   language="Hausa", max_words=60))
```
